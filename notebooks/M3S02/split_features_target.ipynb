# [M3S02] - Ex. 1 — Separação de X e y

**Objetivo:** separar as variáveis preditoras (X) da variável resposta (y) a partir da base de dados farmacêutica.

**Base de dados:** `dados_AP - [M3S01] - Ex. 1 até [M3S03] - Ex. 4.csv`

```python
import pandas as pd
```

```python
df = pd.read_csv('../../data/input/csv/dados_AP - [M3S01] - Ex. 1 até [M3S03] - Ex. 4.csv')
df.head()
```

## Separação de X e y

```python
X = df[['investimento_marketing', 'preco_medicamento', 'mes']]
y = df['demanda']

print('Variáveis preditoras (X):')
print(f'  Colunas : {list(X.columns)}')
print(f'  Shape   : {X.shape}')
print()
print('Variável resposta (y):')
print(f'  Coluna  : {y.name}')
print(f'  Shape   : {y.shape}')
```

## Conclusão

A separação de **X** e **y** é o primeiro passo obrigatório antes de treinar qualquer modelo de ML.

| | Variáveis | Tipo |
|---|---|---|
| **X** | `investimento_marketing`, `preco_medicamento`, `mes` | DataFrame (entrada do modelo) |
| **y** | `demanda` | Series (o que o modelo deve prever) |

Essa estrutura é compatível com a API do **scikit-learn**, onde todos os estimadores recebem `model.fit(X, y)`.
